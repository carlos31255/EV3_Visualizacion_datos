# 02 — EDA y visualizaciones: películas vs series
 **Caso:** StreamView Analytics

**Pregunta de negocio:** ¿cómo debería StreamView repartir su inversión en adquisición entre películas y series, y en qué géneros/países?

**Audiencia (todos los gráficos):** Gerencia de Contenido y Dirección Ejecutiva.

**Reglas:** una métrica = un criterio (orden = largo = título) · porcentajes sobre base fija · Película = azul, Serie = naranja · gris para lo no destacado.

**Definición de calificación:** G1 y G5 promedian cualquier título con votos (junto al % con ≥ 10 votos); G3 y G7 usan solo títulos con ≥ 10 votos. Un título cuenta en cada género y país que lista; el dashboard (Looker) usa solo el género, el país y el director principal (ver controles al final de G8).

**Hilo del análisis:** G1, G2 y G5 miden cuánto confiar en la nota y en la popularidad; G3, G4, G6 y G7 muestran qué géneros y países concentran la oferta y si lo más popular se califica mejor; G8 explora la relación entre recaudación y presupuesto. La síntesis responde la pregunta de negocio.

**Alcance EV3:** se conserva la base validada en EV2. `indice_engagement` es popularidad relativa dentro de cada tipo, no reproducciones ni engagement real. El umbral de 10 votos es un criterio operativo y no garantiza representatividad estadística.


```python
import sys
import textwrap
from pathlib import Path
from typing import cast

import numpy as np
import pandas as pd
import matplotlib as mpl
import matplotlib.pyplot as plt
from IPython.display import Markdown, display


# Localizar la raíz del proyecto.
ROOT = next(
    (
        ruta
        for ruta in [Path.cwd(), *Path.cwd().parents]
        if (ruta / "src" / "limpieza.py").is_file()
        and (ruta / "data" / "processed").is_dir()
    ),
    None,
)

if ROOT is None:
    raise FileNotFoundError(
        "Ejecuta desde la raíz del proyecto o notebooks/: "
        "se necesitan src/ y data/processed/."
    )

# Cargar módulos locales.
ruta_src = str(ROOT / "src")

if ruta_src not in sys.path:
    sys.path.insert(0, ruta_src)

from kpis import PAISES_ES


# Rutas y archivos necesarios.
PROC = ROOT / "data" / "processed"
IMG = ROOT / "images"
IMG.mkdir(parents=True, exist_ok=True)

requeridos = [
    "contenido_unificado.csv",
    "contenido_por_genero.csv",
    "contenido_por_pais.csv",
    "contenido_por_director.csv",
    "looker_contenido.csv",
]

faltan = [
    nombre
    for nombre in requeridos
    if not (PROC / nombre).is_file()
]

if faltan:
    raise FileNotFoundError(
        f"Faltan {faltan}. Ejecuta src/limpieza.py "
        "y después src/kpis.py desde la raíz."
    )


# Configuración visual.
TIPOS = ["Película", "Serie"]
COLOR = {
    "Película": "#0072B2",
    "Serie": "#E69F00",
}
GRIS, TEXTO = "#BDBDBD", "#222222"

mpl.rcParams.update({
    "font.size": 10,
    "axes.edgecolor": "#888888",
    "text.color": TEXTO,
    "axes.labelcolor": TEXTO,
    "xtick.color": TEXTO,
    "ytick.color": TEXTO,
    "axes.spines.top": False,
    "axes.spines.right": False,
    "axes.grid": False,
})


# Cargar datos.
df = pd.read_csv(PROC / "contenido_unificado.csv")
gen = pd.read_csv(PROC / "contenido_por_genero.csv")
pais = pd.read_csv(PROC / "contenido_por_pais.csv")
direc = pd.read_csv(PROC / "contenido_por_director.csv")

NO_ESP = "No especificado"

# Denominador fijo para porcentajes sobre el total de cada tipo.
N_TIPO = df.groupby("tipo_contenido").size()

# Cifras clave para controles cruzados.
CIFRAS = []


# Funciones compartidas por los gráficos.
def encabezado(fig, mensaje, subtitulo):
    ancho = max(65, int(fig.get_figwidth() * 12))

    nota = "\n".join(
        textwrap.fill(linea, width=ancho)
        for linea in subtitulo.splitlines()
    )

    titulo = textwrap.fill(
        mensaje,
        width=max(60, int(fig.get_figwidth() * 9)),
    )

    fig.text(
        0.01, 1.02, nota,
        ha="left",
        va="bottom",
        fontsize=9,
        color="#555555",
    )

    fig.text(
        0.01,
        1.02 + (nota.count("\n") + 1) * 0.035 + 0.025,
        titulo,
        ha="left",
        va="bottom",
        fontsize=13,
        fontweight="bold",
    )


def guardar(fig, nombre):
    fig.savefig(
        IMG / nombre,
        dpi=200,
        bbox_inches="tight",
        facecolor="white",
    )
    plt.show()
    plt.close(fig)


def cifra(grafico, nombre, valor):
    CIFRAS.append({
        "grafico": grafico,
        "cifra": nombre,
        "valor": round(float(valor), 3),
    })
```

## G1 — Calidad y cobertura de votos
- **Problema:** ¿qué tipo de contenido se percibe mejor, y cuánto podemos confiar en esa nota?
- **Mensaje:** las series puntúan más alto, pero pocas tienen votos suficientes.
- **Métrica:** calificación promedio (solo con votos) y % con ≥ 10 votos (base: total del tipo).
- **Por qué barras:** dos valores por tipo; la longitud desde cero se compara sin esfuerzo.

```python
q = df.groupby("tipo_contenido").agg(calif=("vote_average_clean", "mean"),
                                     conf=("calificacion_confiable", "mean"))
q["conf"] *= 100
fig, ax = plt.subplots(1, 2, figsize=(10, 4.2))
for a, col, fmt, lim, ttl in [
        (ax[0], "calif", "{:.2f}", 10, "Calificación promedio (0–10), solo títulos con votos"),
        (ax[1], "conf", "{:.1f} %", 100, "% de títulos con ≥ 10 votos (base: total del tipo)")]:
    vals = q.loc[TIPOS, col]
    b = a.bar(TIPOS, vals, color=[COLOR[t] for t in TIPOS], width=0.55)
    a.bar_label(b, labels=[fmt.format(v) for v in vals], padding=3, fontweight="bold")
    a.set_ylim(0, lim); a.set_yticks([]); a.spines["left"].set_visible(False)
    a.set_title(ttl, loc="left", fontsize=10)
    calif_serie = cast(float, q.loc["Serie", "calif"])
    calif_pelicula = cast(float, q.loc["Película", "calif"])

mas = "más" if calif_serie > calif_pelicula else "menos"
encabezado(fig, f"Las series puntúan {mas} alto ({q.loc['Serie','calif']:.2f} vs {q.loc['Película','calif']:.2f}), "
                f"pero solo {q.loc['Serie','conf']:.0f} % tiene votos suficientes",
           "Calificación promedio excluye títulos sin votos (vote_count = 0). Barras desde cero.")
guardar(fig, "g1_calidad_y_cobertura.png")
for t in TIPOS:
    cifra("G1", f"calificacion_prom_{t}", q.loc[t, "calif"]); cifra("G1", f"pct_confiable_{t}", q.loc[t, "conf"])
```

## G2 — Popularidad sesgada: la mediana es más representativa que el promedio
- **Problema:** ¿el promedio de popularidad representa a un título típico?
- **Mensaje:** la distribución es muy sesgada; hay que reportar la mediana.
- **Métrica:** `popularity` original, dentro de cada tipo (escalas no comparables entre tipos).
- **Por qué boxplot:** resume mediana, dispersión y atípicos; los atípicos (> Q3 + 1,5·IQR) se excluyen del dibujo y se cuantifican.

```python
fig, axes = plt.subplots(1, 2, figsize=(9.5, 4.6))
ratios = {}
for ax_, t in zip(axes, TIPOS):
    x = df.loc[df.tipo_contenido == t, "popularity"]
    q1, q3 = x.quantile([0.25, 0.75]); valla = q3 + 1.5 * (q3 - q1)
    n_out = int((x > valla).sum()); ratios[t] = x.mean() / x.median()
    # Boxplot desplazado a la izquierda para dejar espacio al texto
    ax_.boxplot(x, positions=[0.6], widths=0.35, showfliers=False, whis=1.5, patch_artist=True,
                boxprops=dict(facecolor=COLOR[t], alpha=0.85),
                medianprops=dict(color="black", linewidth=2))
    ax_.set_xlim(0.2, 1.5)
    ax_.set_xticks([]); ax_.set_title(t + "s", loc="left", fontsize=10, fontweight="bold")
    ax_.set_ylabel("Popularidad (valor original de la fuente)")
    
    # Texto a la derecha, separado del boxplot
    ax_.text(0.98, 0.85, f"Mediana: {x.median():.1f}\nPromedio: {x.mean():.1f}\n\n"
             f"Atípicos (> Q3 + 1,5·IQR):\n{n_out:,} títulos\n({n_out/len(x)*100:.1f} % del tipo,\nfuera del gráfico)",
             transform=ax_.transAxes, ha="right", va="center", fontsize=8.5,
             color="#333333", linespacing=1.3)
    cifra("G2", f"mediana_popularidad_{t}", x.median()); cifra("G2", f"atipicos_{t}", n_out)
encabezado(fig, f"La popularidad está muy sesgada: el promedio es {ratios['Película']:.1f}× la mediana en películas "
                f"y {ratios['Serie']:.1f}× en series",
           "Cada panel tiene su propia escala (no comparar alturas entre tipos). Atípicos excluidos con regla IQR; "
           "por eso se reporta la mediana.")
fig.tight_layout(); guardar(fig, "g2_popularidad_boxplot_iqr.png")
```

## G3 — Popularidad relativa y calidad por género: los géneros más populares no siempre son los mejor calificados
- **Problema:** ¿qué géneros combinan alta popularidad relativa y buena calificación?
- **Mensaje:** el título del gráfico indica qué género lidera cada métrica en cada tipo (se calcula desde los datos).
- **Métrica:** eje X = popularidad relativa mediana (percentil dentro del tipo); eje Y = calificación promedio solo con ≥ 10 votos; tamaño = cantidad de títulos.
- **Por qué dispersión:** dos métricas continuas por género; la posición es el atributo más fácil de leer. Líneas punteadas = percentil 50 y calificación media del tipo con ≥10 votos. Para evitar la sobrecarga cognitiva y la superposición de textos, solo se etiquetan los géneros más extremos (aquellos más alejados del promedio) y el más voluminoso, además de Documental, Ciencia ficción y Fantasía y Acción y Aventura, dejando los puntos centrales como contexto.

- **Bases:** la mediana de popularidad usa todos los títulos del género; la nota usa títulos con ≥10 votos. Se exige ≥100 títulos y ≥30 notas válidas. La línea vertical marca el percentil 50; la horizontal, la nota promedio del tipo con ≥10 votos. No representan el mismo estadístico.


```python
g = gen[gen.genero_unificado != NO_ESP]                       # también lo usa G4
base = g.assign(nota=g.vote_average_clean.where(g.calificacion_confiable))   # solo notas con ≥ 10 votos
d = (base.groupby(["tipo_contenido", "genero_unificado"])
         .agg(n=("content_id", "size"), eng=("indice_engagement", "median"),
              calif=("nota", "mean"), n_conf=("nota", "count"))
         .reset_index())
d = d[(d.n >= 100) & (d.n_conf >= 30)]

fig, axes = plt.subplots(1, 2, figsize=(13, 6), sharey=True)
lider = {}
for ax_, t in zip(axes, TIPOS):
    s = d[d.tipo_contenido == t]
    lider_eng = s.nlargest(1, "eng")["genero_unificado"].iloc[0]
    lider_calif = s.nlargest(1, "calif")["genero_unificado"].iloc[0]

    lider[t] = (lider_eng, lider_calif)
    ax_.scatter(s.eng, s.calif, s=s.n / s.n.max() * 700 + 40, color=COLOR[t], alpha=0.75, edgecolor="white")
    # rotular solo los 6 géneros más alejados del valor típico y el más grande (evita que se pisen las etiquetas)
    dist = ((s.eng - 50) / s.eng.std()) ** 2 + ((s.calif - s.calif.mean()) / s.calif.std()) ** 2
    rotular = set(s.loc[dist.nlargest(6).index, "genero_unificado"]) | set(s.loc[s.n.idxmax():s.n.idxmax(), "genero_unificado"])
    rotular |= {"Documental", "Ciencia ficción y Fantasía", "Acción y Aventura"}
    # Etiquetas por panel, con líneas guía: evita solapamientos entre géneros.
    posiciones = {
        "Película": {"Drama": (-12, 14, "right"),
                    "Acción y Aventura": (-12, -28, "right"),
                    "Ciencia ficción y Fantasía": (-12, 18, "right")},
        "Serie": {"Documental": (-18, 30, "right"),
                  "Ciencia ficción y Fantasía": (22, 24, "left"),
                  "Acción y Aventura": (65, 0, "left"),
                  "Drama": (-20, -24, "right")}
    }
    ax_.margins(x=0.30, y=0.18)
    for _, r_ in s[s.genero_unificado.isin(rotular)].iterrows():
        xt, yt, ha = posiciones[t].get(r_.genero_unificado, (10, 12, "left"))
        etiqueta = r_.genero_unificado.replace("Ciencia ficción y Fantasía", "Ciencia ficción\ny Fantasía")
        ax_.annotate(etiqueta, (r_.eng, r_.calif), xytext=(xt, yt),
                     textcoords="offset points", fontsize=8.5, ha=ha, va="center",
                     bbox=dict(facecolor="white", edgecolor="none", alpha=0.85, pad=1.5),
                     arrowprops=dict(arrowstyle="-", color="#777777", lw=0.7),
                     annotation_clip=False)
    ax_.axvline(50, color=GRIS, linewidth=1, linestyle="--")      # referencia de posición relativa: percentil 50
    ax_.axhline(df.loc[(df.tipo_contenido == t) & df.calificacion_confiable, "vote_average_clean"].mean(),
                color=GRIS, linewidth=1, linestyle="--")
    ax_.set_title(f"{t}s", loc="left", fontsize=10, fontweight="bold")
    ax_.set_xlabel("Popularidad relativa mediana (percentil del tipo)")
    cifra("G3", f"eng_max_{t}", s.eng.max()); cifra("G3", f"calif_max_{t}", s.calif.max())
axes[0].set_ylabel("Calificación promedio (solo ≥ 10 votos)")
mejor = (f"{lider['Película'][1]} en ambos tipos" if lider['Película'][1] == lider['Serie'][1]
         else f"{lider['Película'][1]} (películas) y {lider['Serie'][1]} (series)")
encabezado(fig, f"Mayor popularidad relativa: {lider['Película'][0]} (películas) y {lider['Serie'][0]} (series); mejor nota: {mejor}",
           "Géneros con ≥100 títulos y ≥30 notas con ≥10 votos; tamaño = títulos. "
           "X: todos los títulos del género; Y: solo ≥10 votos. Líneas: percentil 50 y nota media del tipo. Popularidad relativa, no consumo absoluto.")
fig.tight_layout(); guardar(fig, "g3_engagement_calidad_genero.png")
```

## G4 — Drama domina ambos catálogos: la oferta está concentrada en pocos géneros
- **Problema:** ¿dónde está concentrada la oferta actual de cada tipo?
- **Mensaje:** Drama es el género más frecuente en películas (43 %) y en series (49 %). Si cambian los datos, actualizar con `cifras_clave_eda.csv`.
- **Métrica:** % de títulos sobre el total del tipo (base fija; un título puede tener varios géneros, por eso no suma 100 %).
- **Por qué barras agrupadas:** comparación directa película vs serie por género, top 7.

```python
comunes = (
    set(g.loc[g["tipo_contenido"] == "Película", "genero_unificado"])
    & set(g.loc[g["tipo_contenido"] == "Serie", "genero_unificado"])
)

v = (
    g[g["genero_unificado"].isin(comunes)]
    .groupby(["genero_unificado", "tipo_contenido"])
    .size()
    .unstack(fill_value=0)
)

# Porcentajes sobre el total de títulos de cada tipo.
v = v.div(N_TIPO[v.columns], axis=1) * 100

# Top 7 según la suma de los porcentajes de ambos tipos.
v = v.loc[v.sum(axis=1).nlargest(7).index].iloc[::-1]

fig, ax = plt.subplots(figsize=(8, 4.8))

h = 0.38
y = np.arange(len(v))

for i, t in enumerate(TIPOS):
    barras = ax.barh(
        y + (i - 0.5) * h,
        v[t].to_numpy(dtype=float),
        height=h,
        color=COLOR[t],
        label=t,
    )
    ax.bar_label(
        barras,
        fmt="%.0f %%",
        padding=3,
        fontsize=8.5,
    )

ax.set_yticks(y, labels=v.index.tolist())
ax.set_xticks([])
ax.spines["bottom"].set_visible(False)
ax.legend(frameon=False, loc="lower right")

# Género líder de cada tipo.
a_ = str(v["Película"].idxmax())
b_ = str(v["Serie"].idxmax())

# Porcentajes del género a_ en ambos tipos.
pct_pelicula = cast(float, v.at[a_, "Película"])
pct_serie = cast(float, v.at[a_, "Serie"])

msg = (
    f"{a_} domina ambos catálogos "
    f"({pct_pelicula:.0f} % de películas, {pct_serie:.0f} % de series)"
    if a_ == b_
    else f"{a_} es el género más frecuente en películas y {b_} en series"
)

encabezado(
    fig,
    msg,
    f"% sobre el total de títulos de cada tipo "
    f"({N_TIPO['Película']:,} y {N_TIPO['Serie']:,}); "
    "un título puede tener varios géneros. "
    "Top 7 de los géneros presentes en ambos tipos.",
)

guardar(fig, "g4_volumen_por_genero.png")

cifra("G4", f"pct_{a_}_Película", pct_pelicula)
cifra("G4", f"pct_{a_}_Serie", pct_serie)
```

## G5 — Los títulos de 2025 tienen menor cobertura de votos en la muestra
- **Problema:** ¿la calidad percibida cambia con los años, y es confiable para los títulos recientes?
- **Mensaje:** hasta 2024 la cobertura de votos fue alta en películas y creciente en series; la caída se concentra en 2025 en la extracción disponible. No se confirma su causa ni se interpreta como una actualización al año actual.
- **Métrica:** calificación promedio (solo con votos) y % con ≥ 10 votos, por año.
- **Por qué líneas:** tendencia temporal; etiquetas directas al final de cada línea y franja gris en el año reciente. Hay 1.000 títulos por año y tipo (991 en series de 2025), así que el volumen por año no es un hallazgo.

```python
a = (df.groupby(["release_year", "tipo_contenido"])
       .agg(calif=("vote_average_clean", "mean"), conf=("calificacion_confiable", "mean")).reset_index())
a["conf"] *= 100
y1 = int(a.release_year.max())
c1 = a[a.release_year == y1].set_index("tipo_contenido").conf
c0 = a[a.release_year == y1 - 1].set_index("tipo_contenido").conf
y0 = int(a.release_year.min())
s0 = a[(a.release_year == y0) & (a.tipo_contenido == "Serie")].conf.iloc[0]
pmin = a[(a.tipo_contenido == "Película") & (a.release_year < y1)].conf.min()
fig, axes = plt.subplots(1, 2, figsize=(10, 4.2), sharex=True)
for ax_, col, ttl, ylim in [(axes[0], "calif", "Calificación promedio (solo con votos)", None),
                            (axes[1], "conf", "% de títulos con ≥ 10 votos", (0, 100))]:
    ax_.axvspan(y1 - 0.5, y1 + 0.5, color=GRIS, alpha=0.3, linewidth=0)   # año reciente
    for t in TIPOS:
        s = a[a.tipo_contenido == t]
        ax_.plot(s.release_year, s[col], color=COLOR[t], linewidth=2.2, marker="o", markersize=3.5)
        y_et = {"Película": 19, "Serie": 6}[t] if col == "conf" else s[col].iloc[-1]   # evita que las etiquetas se pisen
        ax_.text(y1 + 0.8, y_et, f"{t}\n{s[col].iloc[-1]:.1f}",
                 color=COLOR[t], fontsize=8.5, fontweight="bold", va="center")
    ax_.set_title(ttl, loc="left", fontsize=10)
    if ylim: ax_.set_ylim(*ylim)
    ax_.set_xlim(a.release_year.min(), y1 + 3.2)
    ax_.set_xticks(range(int(a.release_year.min()), y1 + 1, 3))
encabezado(fig, f"En {y1} pocos títulos alcanzan ≥10 votos en la muestra "
                f"(series {c1['Serie']:.0f} %, películas {c1['Película']:.0f} %)",
           f"Hasta {y1 - 1}: películas siempre ≥ {pmin:.0f} % y series en alza ({s0:.0f} % en {y0} → {c0['Serie']:.0f} % en {y1 - 1}); "
           f"el desplome es solo de {y1} (franja gris). Datos de la extracción original; 1.000 títulos/año y tipo, salvo 991 series en 2025.")
fig.tight_layout(); guardar(fig, "g5_evolucion_por_anio.png")
cifra("G5", f"conf_series_{y1 - 1}", c0["Serie"]); cifra("G5", f"conf_series_{y1}", c1["Serie"])
cifra("G5", f"conf_peliculas_{y1}", c1["Película"])
```

## G6 — Estados Unidos domina las películas, pero no las series
- **Problema:** ¿de dónde viene la oferta de cada tipo?
- **Mensaje:** el país líder concentra casi la mitad de las películas y solo una quinta parte de las series, donde China y Japón pesan casi lo mismo entre sí.
- **Métrica:** % de títulos sobre el total del tipo (base fija), top 7 por tipo, sin «No especificado»; países en español.
- **Por qué barras:** ranking de categorías con nombres largos.
- **Nota sobre Looker:** G6 cuenta todos los países de producción de cada título; el dashboard usa solo `pais_principal`, por lo que EE. UU. baja de 48,5 % a 35,6 % en películas y de 20,0 % a 15,5 % en series (control en la celda posterior al de G4).

```python
if ROOT is None:
    raise FileNotFoundError(
        "No se encontró la raíz del proyecto. Ejecuta primero la celda inicial."
    )

sys.path.append(str(ROOT / "src"))

# traducción única de países (la misma que usa Looker)
p = pais[pais.pais != NO_ESP].assign(pais=lambda x: x.pais.map(PAISES_ES).fillna(x.pais))
fig, axes = plt.subplots(1, 2, figsize=(10, 4.4))
share = {}
for ax_, t in zip(axes, TIPOS):
    s = (p[p.tipo_contenido == t].groupby("pais").size() / N_TIPO[t] * 100)
    share[t] = s
    s7 = s.nlargest(7).sort_values()
    b = ax_.barh(s7.index, s7.values, color=COLOR[t])
    ax_.bar_label(b, fmt="%.1f %%", padding=3, fontsize=8.5)
    ax_.set_xticks([]); ax_.spines["bottom"].set_visible(False)
    ax_.set_title(t + "s (top 7)", loc="left", fontsize=10, fontweight="bold")
    cifra("G6", f"pct_{s7.index[-1]}_{t}", s7.values[-1])
top = share["Película"].idxmax()
otros = share["Serie"].drop(top).nlargest(2)
encabezado(fig, f"{top} concentra {share['Película'][top]:.0f} % de las películas y solo {share['Serie'][top]:.0f} % de las series",
           f"En series, {otros.index[0]} ({otros.iloc[0]:.0f} %) y {otros.index[1]} ({otros.iloc[1]:.0f} %) pesan casi lo mismo entre sí. "
           f"% sobre el total de títulos de cada tipo (base fija); un título puede tener varios países. Excluye «{NO_ESP}».")
fig.tight_layout(); guardar(fig, "g6_paises.png")
```

## G7 — Relación débil: los títulos más populares tienden a tener algo mejor nota
- **Problema:** ¿los títulos más populares son también los mejor calificados?
- **Mensaje:** la correlación lineal observada es débil (r = 0,25 en películas y 0,13 en series). No demuestra causalidad ni permite concluir independencia.
- **Métrica:** `indice_engagement` (percentil dentro del tipo) vs `vote_average_clean`, solo títulos con ≥ 10 votos.
- **Por qué hexbin:** miles de puntos se superponen en un scatter; la densidad por hexágono (escala log rotulada) evita el amontonamiento.

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4.6), sharey=True)
rs = {}
for ax_, t, cm in zip(axes, TIPOS, ["Blues", "Oranges"]):
    s = df[(df.tipo_contenido == t) & df.calificacion_confiable]
    rs[t] = s.indice_engagement.corr(s.vote_average_clean)
    hb = ax_.hexbin(s.indice_engagement, s.vote_average_clean, gridsize=28, mincnt=1, bins="log", cmap=cm)
    plt.colorbar(hb, ax=ax_, label="Títulos (escala log)")
    ax_.set_xlabel("Popularidad relativa (percentil del tipo, 0–100)")
    ax_.set_title(f"{t}s · n = {len(s):,} · r = {rs[t]:.2f}", loc="left", fontsize=10, fontweight="bold")
axes[0].set_ylabel("Calificación (0–10)")
rm = max(abs(r) for r in rs.values())
rel = "muy débil" if rm < 0.1 else "débil" if rm < 0.4 else "apreciable"
encabezado(fig, f"Popularidad relativa y calificación: asociación lineal {rel} (r = {rs['Película']:.2f} películas, {rs['Serie']:.2f} series)",
           "Solo títulos con ≥ 10 votos. Hexágonos más oscuros = más títulos (escala log, rotulada).")
fig.tight_layout(); guardar(fig, "g7_engagement_vs_calidad.png")
for t in TIPOS: cifra("G7", f"r_{t}", rs[t])
```

## G8 — Recaudación / presupuesto por director (solo películas): ¿el liderazgo se apoya en una sola película?
- **Problema:** ¿qué directores presentan mayor recaudación por unidad de presupuesto registrado, y cuánto depende ese ranking de un único éxito?
- **Mensaje:** el título del gráfico dice si el director líder depende de una sola película (≥ 50 % de sus ingresos) o no.
- **Métrica:** múltiplo agregado = Σ ingresos ÷ Σ presupuesto, directores con ≥ 5 películas con datos. Orden, largo y título = múltiplo agregado; la mediana y la dependencia se rotulan como apoyo. Se unifica a la dupla Anthony y Joe Russo para evitar filas duplicadas con idénticas métricas.
- **Por qué barras con línea de 1×:** ranking con referencia recaudación = presupuesto. Las series no tienen datos financieros.

- **Criterio de director:** en este gráfico exploratorio cada director cuenta con todas sus películas, incluso las codirigidas, porque la pregunta es qué directores rinden más. El dashboard asigna cada película a un solo director (el primero listado) para que los ingresos no se cuenten dos veces y el ROI global (2,92×) y la tabla por director salgan de la misma base. Por eso Chris Renaud aparece aquí con 6 películas y 8,8×, y en Looker con 5 películas y 8,96×. Además, aquí el top 10 se ordena por ROI y en Looker por ingresos.

**Límite:** el cociente recaudación/presupuesto no es rentabilidad neta: faltan otros costos y el reparto de ingresos. Se conserva `roi` como nombre técnico por compatibilidad.


```python
MIN_PELIS = 5                                   # mínimo de películas con datos por director
f = direc[(direc.tipo_contenido == "Película") & direc.tiene_datos_financieros].copy()
# Unificar a Anthony y Joe Russo como dupla para evitar filas duplicadas con idénticas métricas
f["director_ind"] = f["director_ind"].replace({
    "Anthony Russo": "Anthony y Joe Russo",
    "Joe Russo": "Anthony y Joe Russo"
})
f = f.drop_duplicates(subset=["content_id", "director_ind"])

r = (f.groupby("director_ind").agg(n=("content_id", "nunique"), bud=("budget_clean", "sum"),
                                  rev=("revenue_clean", "sum"), mediana=("roi", "median"),
                                  rev_max=("revenue_clean", "max")).reset_index())
r["roi_agregado"] = r.rev / r.bud
r["dep"] = r.rev_max / r.rev * 100              # % de los ingresos que aporta su película más taquillera
r = r[r.n >= MIN_PELIS].nlargest(10, "roi_agregado").sort_values("roi_agregado")
fig, ax = plt.subplots(figsize=(10, 4.8))
ax.barh(r.director_ind, r.roi_agregado,
        color=[COLOR["Película"] if i == len(r) - 1 else GRIS for i in range(len(r))])
for yy, (v_, m_, n_, d_) in enumerate(zip(r.roi_agregado, r.mediana, r.n, r.dep)):
    ax.text(v_, yy, f" {v_:.1f}× · mediana {m_:.1f}× · {n_} pelis · top película = {d_:.0f} % de ingresos",
            va="center", fontsize=8)
ax.axvline(1, color="black", linewidth=1, linestyle="--")
ax.text(1, len(r) - 0.4, " recaudación = presupuesto (1×)", fontsize=8)
ax.set_xticks([]); ax.spines["bottom"].set_visible(False); ax.set_xlim(0, r.roi_agregado.max() * 2.1)
cob = df.loc[df.tipo_contenido == "Película", "tiene_datos_financieros"].mean() * 100
roi_med_global = df.loc[df.tiene_datos_financieros, "roi"].median()    # múltiplo mediano por película (1,70×); no es el múltiplo agregado de la tarjeta de D4 (Σ ingresos ÷ Σ presupuesto, 2,92×)
lid = r.iloc[-1]
frase = (f"una sola película aporta {lid.dep:.0f} % de sus ingresos" if lid.dep >= 50
         else f"sus ingresos están repartidos (la mayor película aporta {lid.dep:.0f} %)")
encabezado(fig, f"{lid.director_ind} lidera el múltiplo agregado ({lid.roi_agregado:.1f}×); {frase}",
           f"múltiplo agregado = Σ ingresos ÷ Σ presupuesto, directores con ≥ {MIN_PELIS} películas con datos; se rotula también la mediana por director.\n"
           f"Múltiplo bruto, no rentabilidad neta. Russo unificados. Mediana global con datos: {roi_med_global:.2f}×. Cobertura: {cob:.1f} % de las {N_TIPO['Película']:,} películas.")
guardar(fig, "g8_roi_por_director.png")
cifra("G8", "roi_agregado_top1", lid.roi_agregado); cifra("G8", "cobertura_financiera_pct", cob)
cifra("G8", "roi_mediano_global", roi_med_global); cifra("G8", "dependencia_top1_pct", lid.dep)
```

## Controles de consistencia con el dashboard
G4, G6 y G8 cuentan cada título en todos sus géneros, países o directores; Looker usa solo el principal (el primero listado) para no duplicar títulos. Las celdas siguientes cuantifican esa diferencia y guardan el resultado en `data/processed/`.

```python
# Control: G4 cuenta todos los géneros de cada título; Looker filtra solo por genero_principal.
lk = pd.read_csv(PROC / "looker_contenido.csv")
n_lk = lk.groupby("tipo_contenido").size()
lkp = (lk[lk.genero_principal != NO_ESP].groupby(["genero_principal", "tipo_contenido"]).size()
         .unstack(fill_value=0).div(n_lk, axis=1) * 100)
g4 = (g[g.genero_unificado.isin(comunes)].groupby(["genero_unificado", "tipo_contenido"]).size()
        .unstack(fill_value=0).div(N_TIPO, axis=1) * 100)
cmp_ = (g4.join(lkp, lsuffix="_G4_todos_los_generos", rsuffix="_Looker_genero_principal", how="left").round(1))
cmp_.to_csv(PROC / "control_g4_vs_looker.csv")
display(cmp_.sort_values(cmp_.columns[0], ascending=False).head(7))
print("Diferencia esperada: G4 cuenta cada género de un título; Looker solo el primero. Declararlo en la página de limitaciones.")
```

```python
# Control: G6 cuenta todos los países de cada título; Looker filtra solo por pais_principal.
# (usa lk, n_lk y p definidos en las celdas anteriores)
lkp_pais = (lk[lk.pais_principal != NO_ESP].groupby(["pais_principal", "tipo_contenido"]).size()
              .unstack(fill_value=0).div(n_lk, axis=1) * 100)
g6 = (p.groupby(["pais", "tipo_contenido"]).size()
        .unstack(fill_value=0).div(N_TIPO, axis=1) * 100)
cmp_p = (g6.join(lkp_pais, lsuffix="_G6_todos_los_paises", rsuffix="_Looker_pais_principal", how="left").round(1))
cmp_p.to_csv(PROC / "control_g6_vs_looker.csv")
display(cmp_p.sort_values(cmp_p.columns[0], ascending=False).head(7))
print("Diferencia esperada: G6 cuenta cada país de un título; Looker solo el primero. Declararlo en la página de limitaciones.")
```

```python
# Control: G8 cuenta cada director en todas sus películas; Looker usa solo el primer director listado.
lk_f = lk[(lk.financiero_num == 1) & (lk.director_principal != NO_ESP)]
c = (lk_f.groupby("director_principal")
         .agg(n_Looker=("content_id", "size"), ing=("ingresos_roi", "sum"), pres=("presupuesto_roi", "sum")))
c["roi_Looker"] = c.ing / c.pres
c = c[c.n_Looker >= MIN_PELIS].nlargest(10, "ing").rename(index={"Anthony Russo / Joe Russo": "Anthony y Joe Russo"})
g8_all = f.groupby("director_ind").agg(n_G8=("content_id", "nunique"), bud=("budget_clean", "sum"), rev=("revenue_clean", "sum"))
g8_all["roi_G8"] = g8_all.rev / g8_all.bud
cmp_d = c.join(g8_all[["n_G8", "roi_G8"]], how="left").round(2)
cmp_d.to_csv(PROC / "control_g8_vs_looker.csv")
display(cmp_d[["n_Looker", "n_G8", "roi_Looker", "roi_G8"]])
print("Diferencia esperada: G8 cuenta a cada director en todas sus películas; Looker solo el primero listado.")
```
